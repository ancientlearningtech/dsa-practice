# Problem 1 — Binary Search Fundamentals

You are given a sorted array of integers numbers and an integer target. Return the index of target if it exists in the array. If target does not exist, return -1

Exmple:

    numbers = [1, 3, 5, 7, 9]
    target = 7
    Output:3

```python
def binarySearch(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        if numbers[mid] > target:
            right = mid - 1
        elif numbers[mid] < target:
            left = mid + 1
        else:
            return mid
    
    return -1
    
print(binarySearch([1,3,5,7,9], 7))
print(binarySearch([1,3,5,7,9], 4))
```

# Problem 2 — Find the First Position Where numbers[i] >= target

Given a sorted array of integers numbers and an integer target, return the index of the first element that is greater than or equal to target. If every element is smaller than target, return -1

Example

    numbers = [1, 3, 5, 7, 9]
    target = 5
    Output: 2

```python
def findFirstGreaterOrEqual(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        if numbers[mid] >= target:
            right = mid - 1
        else:
            left = mid + 1
    if left <= len(numbers) - 1:
        return left
    
    return -1

print(findFirstGreaterOrEqual([1, 3, 5, 7, 9], 5)) # 2
print(findFirstGreaterOrEqual([1, 3, 5, 5, 5, 7, 9], 5)) # 2
print(findFirstGreaterOrEqual([1, 3, 5, 5, 5, 7, 9], 6)) # 5
print(findFirstGreaterOrEqual([1, 3, 5, 7, 9], 6)) # 3
print(findFirstGreaterOrEqual([1, 3, 5, 7, 9], 10)) # -1
```

# Problem 3 — Find the Integer Square Root

Given a non-negative integer number, return the integer square root of number. 

The integer square root is the largest integer whose square is less than or equal to number. 

You must solve this using Binary Search, without using built-in square-root functions or exponentiation to calculate the square root.

    Input:  number = 25, Output: 5
    Input:  number = 26, Output: 5
    Input:  number = 8, Output: 2
    Input:  number = 0, Output: 0

```python
def integerSquareRoot(number):
    left = 0
    right = number
    
    while left <= right:
        mid = left + (right - left) // 2
        if mid * mid > number:
            right = mid - 1
        elif mid * mid < number:
            left = mid + 1
        else:
            return mid
    return right
        
print(integerSquareRoot(25)) # 5
print(integerSquareRoot(26)) # 5
print(integerSquareRoot(8)) # 2
print(integerSquareRoot(0)) # 0
```

# Problem 4 — Find the Floor of a Number

You are given a sorted array of distinct integers numbers and an integer target. Find the floor of the target in the array. The floor is the largest element that is less than or equal to the target. Return the index of that element. If no such element exists, return -1.

Example:

    numbers = [1, 3, 5, 7, 9], target = 6
    The elements less than or equal to 6 are 1, 3, and 5. 
    Output = 2

```python
def findFloor(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        # print(left, mid, right)
        if numbers[mid] > target:
            right = mid - 1
        elif numbers[mid] < target:
            left = mid + 1
        else:
            return mid
    
    if left > right:
        return right
    
    return -1

print(findFloor([1,3,5,7,9], 6)) # 2
print(findFloor([1,3,5,7,9], 7)) # 3
print(findFloor([1,3,5,7,9], 0)) # -1
print(findFloor([1,3,5,7,9], 10)) # 4
```

# Problem 5 — Search in a Sorted Array of Unknown Size

You are given a sorted array of distinct integers, but its length is unknown. You can access an element by its index. If an index is outside the array, assume the access returns a special value representing an unavailable element. Return the index of target if it exists; otherwise, return -1.

Example:

    numbers = [2, 5, 8, 12, 16, 23, 38, 45, 51]
    target = 38
    Output: 6

```python
def findSearchRange(numbers, target):
    left = 0
    right = 1
    
    while right < len(numbers) and numbers[right] < target:
        right *= 2
    
    return left, right

# print(findSearchRange([2, 5, 8, 12, 16, 23, 38, 45, 57, 71, 84, 99], 57))
```

# Problem 6 - Find the Position of a Target in a Nearly Sorted Array

A sorted array has been modified so that an element may have moved one position away from where it normally belongs.

Example:

    numbers = [10, 3, 40, 20, 50, 80, 70], target = 40

Return its index.

```python
def searchNearlySorted(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        if mid > 0 and numbers[mid - 1] == target:
            return mid - 1
        if numbers[mid] == target:
            return mid
        if mid < len(numbers) - 1 and numbers[mid + 1] == target:
            return mid + 1
        if numbers[mid] > target:
            right  = mid - 2
        else:
            left = mid + 2
    
    return -1

print(searchNearlySorted([10, 3, 40, 20, 50, 80, 70], 40))
print(searchNearlySorted([10, 3, 40, 20, 50, 80, 70], 3))
```

# Problem 7 - Find the Rotation Count

Given a sorted array that has been rotated, find how many times it was rotated.

Example:

    Original:
    [1, 2, 3, 4, 5, 6, 7]

    Rotated:
    [4, 5, 6, 7, 1, 2, 3]

    Output: 4

```python
def findRotationCount(numbers):
    left = 0
    right = len(numbers) - 1
    
    while left < right:
        mid = left + (right - left) // 2
        if numbers[mid] > numbers[right]:
            left = mid + 1
        else:
            right = mid
    return left

print(findRotationCount([4,5,6,7,1,2,3]))
# print(findRotationCount([5,6,7,1,2,3,4]))
# print(findRotationCount([1, 2, 3, 4, 5]))
# print(findRotationCount([5, 6, 7, 1, 2, 3, 4]))
# print(findRotationCount([6, 7, 1, 2, 3, 4, 5]))
print(findRotationCount([ 7, 1, 2, 3, 4, 5, 6]))
```

# Problem 8 - Find the Closest Element

Given a sorted array and a target, find the element whose value is closest to the target.

Example:

    numbers = [1, 3, 5, 7, 9]
    target = 6

The closest values are:

    5 → difference = 1
    7 → difference = 1

If there is a tie, return the smaller value.

So the answer is: 5

```python
def findClosestElement(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        
        if numbers[mid] < target:
            left = mid + 1
        elif numbers[mid] > target:
            right = mid - 1
        else:
            return numbers[mid]
    
    if right < 0:
        return numbers[left]
    if left == len(numbers):
        return numbers[right]
    
    leftDifference = abs(numbers[left] - target)
    rightDifference = abs(numbers[right] - target)
    
    if leftDifference < rightDifference:
        return numbers[left]
    else:
        return numbers[right]

print(findClosestElement([1,3,5,7,9], 6))
print(findClosestElement([1,3,5,7,9], 8))
print(findClosestElement([1,3,5,7,9], 2))
print(findClosestElement([1,3,5,7,9], 10))
```

# Problem 9 - Single Non-Duplicate Element

You are given a sorted array where:

Every element appears exactly twice
- Except for one element, which appears only once
- Find that single element
- The array contains no other duplicates

Example

    numbers = [1, 1, 2, 2, 3, 4, 4, 5, 5], Output: 3

```python
def findSingleElement(numbers):
    left = 0
    right = len(numbers) - 1
    
    while left < right:
        mid = left + (right - left) // 2
        if mid % 2 == 1:
            mid -= 1
        if numbers[mid] != numbers[mid + 1]:
            right = mid
        else:
            left = mid + 2
            
    return numbers[right]

print(findSingleElement([1, 1, 2, 2, 3, 4, 4, 5, 5]))
print(findSingleElement([1, 1, 2, 2, 3, 3, 4, 5, 5]))
```

# Problem 10 - Search in a Bitonic Array

A bitonic array:

- Strictly increases up to one maximum.
- Then strictly decreases.

Example:

    [1, 3, 8, 12, 9, 5, 2]
                ↑
                peak

We need to find a target value.

Given a bitonic array and a target, return the index of the target. If it doesn't exist, return -1.

Example

    numbers = [1, 3, 8, 12, 9, 5, 2]
    target = 9

    Output: 4

```python
def findPeakIndex(numbers):
    left = 0
    right = len(numbers) - 1
    
    while left < right:
        mid = left + (right - left) // 2
        if numbers[mid] > numbers[mid + 1]:
            right = mid
        elif numbers[mid] < numbers[mid + 1]:
            left = mid + 1
        
    return right

def searchBitonicArray(numbers, target):
    left = 0
    peakIndex = findPeakIndex(numbers)
    right = peakIndex
    
    while left <= right:
        mid = left + (right - left) // 2
        if numbers[mid] > target:
            right = mid - 1
        elif numbers[mid] < target:
            left = mid + 1
        else:
            return mid

    left = peakIndex + 1
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        if numbers[mid] > target:
            left = mid + 1
        elif numbers[mid] < target:
            right = mid - 1
        else:
            return mid
        
    return -1
    
print(searchBitonicArray([1,3,5,8,12,9,5,2], 8))
print(searchBitonicArray([1,3,5,8,12,9,5,2], 2))
```
