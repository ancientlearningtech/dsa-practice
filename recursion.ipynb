# Problem 1 - Calculate Factorial Using Recursion

Write a Python function named factorial(number) that returns the factorial of a non-negative integer number using recursion.

Examples
    
    - Input: number = 4 → Output: 24
    - Input: number = 0 → Output: 1

```python
def factorial(number):
    if number == 0:
        return 1
    return number * factorial(number - 1)

print(factorial(3))
```

Examples:

# Problem 4 - Reverse a String Using Recursion
Write a recursive Python function named reverseString(text) that returns the string in reverse order.

Examples:

    - Input: text = "hello" → Output: "olleh"
    - Input: text = "cat" → Output: "tac"

Constraints:

    - 1 <= len(text) <= 1000
    - Do not use a loop.
    - Do not use slicing to reverse the entire string (such as text[::-1]).

```python
def reverseString(text):
    if len(text) == 0:
        return ''
    return text[-1] + reverseString(text[:-1])

print(reverseString('hello'))
```

Examples:
- Input: text = "racecar" → Output: True
- Input: text = "hello" → Output: False
Constraints:
- 1 <= len(text) <= 1000
- Treat uppercase and lowercase letters as different.
- Do not use loops or reverse the entire string.

```python
# O(n) time complexity
def isPalindrome(text):
    characters = list(text)
    
    def helper(left, right):
        if left >= right:
            return True
        if characters[left] != characters[right]:
            return False
        return helper(left + 1, right - 1)
    
    return helper(0, len(characters) - 1)

print(isPalindrome('racecar'))
```

# Problem 6 - Calculate the Power of a Number
Write a recursive Python function named power(base, exponent) that returns base raised to the power exponent, without using Python's built-in ** operator.
Examples
- Input: base = 2, exponent = 4 → Output: 16
- Input: base = 5, exponent = 0 → Output: 1

Constraints
- base is an integer.
- 0 <= exponent <= 1000
- Do not use loops or the ** operator.

```python
def power(base, exponent):
    if exponent == 0:
        return 1
    return base * power(base, exponent - 1)

print(power(2, 4))
```

# Problem 7 - Find the Maximum Number in a List
Write a recursive Python function named findMaximum(numbers) that returns the largest number in a non-empty list, without using max() or loops.
Examples:
- Input: numbers = [3, 7, 2, 9, 4] → Output: 9
- Input: numbers = [-5, -2, -8] → Output: -2

Constraints:
- 1 <= len(numbers) <= 1000
- List elements are integers.
- Do not use loops or max().

```python
def findMaximum(numbers):
    def helper(index):
        if index == 0:
            return numbers[0]
        
        previousMaximum = helper(index - 1)

        if numbers[index] > previousMaximum:
            return numbers[index]
        
        return previousMaximum
    
    return helper(len(numbers) - 1)

print(findMaximum([3,7,2,9,4]))
print(findMaximum([-5,-2,-8]))
```

# Problem 8 - Calculate the Sum of List Elements
Write a recursive Python function named sumList(numbers) that returns the sum of all elements in a list, without using loops or Python's built-in sum() function.
Examples
- Input: numbers = [2, 4, 6, 8] → Output: 20
- Input: numbers = [5] → Output: 5

Constraints
- 1 <= len(numbers) <= 1000
- Elements can be positive, negative, or zero.
- Do not use loops or sum().

```python
def sumList(numbers):
    def helper(index):
        if index == 0:
            return numbers[0]
        return numbers[index] + helper(index - 1)
    return helper(len(numbers) - 1)

print(sumList([2,4,6,8]))
```

# Problem 9 - Fibonacci Number
Write a recursive Python function named fibonacci(number) that returns the Fibonacci number at position number.
The sequence starts with 0, 1, and each subsequent number is the sum of the previous two numbers.
Examples
- Input: number = 5 → Output: 5
- Input: number = 7 → Output: 13

Constraints
- 1 <= number <= 30
- Use recursion; do not use loops.

```python
def fibonacci(number):
    if number == 0:
        return 0
    if number == 1:
        return 1
    return fibonacci(number - 1) + fibonacci(number - 2)

print(fibonacci(7))
```

# Problem 10 - Check Whether a List Is Sorted
Problem statement
Write a recursive Python function named isSorted(numbers) that returns True if the list is sorted in non-decreasing order, and False otherwise.
Examples
- Input: numbers = [1, 2, 2, 4, 7] → Output: True
- Input: numbers = [1, 3, 2, 4] → Output: False

Constraints

- 1 <= len(numbers) <= 1000
- Do not use loops or sorted().

```python
def isSorted(numbers):
    def helper(index):
        if index == 0:
            return True
        
        if numbers[index - 1] > numbers[index]:
            return False
        
        return helper(index - 1)
    
    return helper(len(numbers) - 1)

print(isSorted([1,2,2,4,7])) # True
print(isSorted([1,3,2,4])) # False
```

# Problem 11 - Find the First Occurrence of a Target
Write a recursive Python function named findFirstOccurrence(numbers, target) that returns the index of the first occurrence of target in a list. If the target isn't present, return -1.
Examples
- Input: numbers = [4, 2, 7, 2, 9], target = 2 → Output: 1
- Input: numbers = [1, 3, 5], target = 8 → Output: -1

Constraints

- 0 <= len(numbers) <= 1000
- Elements are integers.
- Do not use loops or list.index().

```python
def findFirstOccurrence(numbers, target):
    def helper(index):
        if index >= len(numbers):
            return -1
        if numbers[index] == target:
            return index
        return helper(index + 1)
        
    return helper(0)
    
print(findFirstOccurrence([9,4,2,7,2,7,9], 9))
```

# Problem 12 - Check Whether a String Is a Subsequence
Write a recursive function isSubsequence(text, target) that returns True if target appears in text in the same order, not necessarily consecutively. Otherwise, return False.

Examples

- text = "abcde", target = "ace" → True
- text = "abcde", target = "aec" → False
- text = "abcde", target = "" → True

Constraints: No loops. Use recursion.

```python
def isSubSequence(text, target):
    def helper(indexOfText, indexOfTarget):
        if indexOfTarget == len(target):
            return True
        
        if indexOfText == len(text):
            return False
        
        if text[indexOfText] == target[indexOfTarget]:
            return helper(indexOfText + 1, indexOfTarget + 1)
        
        return helper(indexOfText + 1, indexOfTarget)
    
    return helper(0, 0)

print(isSubSequence('abcde', 'ace'))
print(isSubSequence('abcde', 'aec'))
print(isSubSequence('abcde', ''))
```

# Problem 13 - Check Whether an Array Is a Palindrome
Write a recursive function named isArrayPalindrome(numbers) that returns True if the array reads the same forward and backward, otherwise False.

Examples
- [1, 2, 3, 2, 1] → True
- [1, 2, 3, 4] → False
- [7] → True

Constraints: Use recursion; no loops.

```python
def isArrayPalindrome(numbers):
    def helper(left, right):
        if left >= right:
            return True
        if numbers[left] != numbers[right]:
            return False    
        return helper(left + 1, right - 1)

    return helper(0, len(numbers) - 1)

print(isArrayPalindrome([1,2,3,2,1]))  
print(isArrayPalindrome([1,2,3,4]))  
print(isArrayPalindrome([7]))        
```

# Problem 14 - Generate Binary Strings
Write a recursive function named generateBinaryStrings(n) that returns all binary strings of length n.
A binary string contains only 0 and 1.

Examples
- n = 2 → ["00", "01", "10", "11"]
- n = 1 → ["0", "1"]
- n = 0 → [""]

Constraints: Use recursion, no loops for generating the strings.

```python
def generateBinaryStrings(n, result):
    if len(result) == n:
        return [result]
    
    zero = generateBinaryStrings(n, result + '0')
    one = generateBinaryStrings(n, result + '1')
    return zero + one

print(generateBinaryStrings(1, ''))
print(generateBinaryStrings(2, ''))
```

# Problem 15 - Count Ways to Climb Stairs
Write a recursive function named countWaysToClimb(n) that returns the number of distinct ways to reach step n when you can climb either 1 or 2 steps at a time.

Examples
- n = 2 → 2 (1+1, 2)
- n = 3 → 3 (1+1+1, 1+2, 2+1)
- n = 4 → 5

Constraint: Use recursion. Try to achieve better than exponential time if you can.

```python
def countWaysToClimb(n):
    def helper(currentSum):
        if currentSum > n:
            return 0
        if currentSum == n:
            return 1
        oneCount = helper(currentSum + 1)
        twoCount = helper(currentSum + 2)
        return oneCount + twoCount
    return helper(0)

print(countWaysToClimb(2))
print(countWaysToClimb(3))
print(countWaysToClimb(4))    
```

# Problem 16 - Sum of Digits Using Recursion
Write a recursive function named sumOfDigits(number) that returns the sum of all digits in a non-negative integer.

Examples
- number = 1234 → 10
- number = 507 → 12
- number = 0 → 0

Constraints: Use recursion, without converting the number into a string or using loops.

```python
def sumOfDigits(number):
    if number < 10:
        return number
    return number % 10 + sumOfDigits(number // 10)

print(sumOfDigits(1234))
print(sumOfDigits(507))
print(sumOfDigits(0))
```

# Problem 17 - Generate All Subsequences of a String
Write a recursive function named generateSubsequences(text) that returns all possible subsequences of a string, including the empty string.
A subsequence preserves character order, but characters don't need to be consecutive.

Examples
- text = "ab" → ["", "b", "a", "ab"] (order may vary)
- text = "a" → ["", "a"]

Constraint: Use recursion, without loops to generate the subsequences.

```python
def generateSubsequences(text):
    def helper(index, currentList):
        if index == len(text):
            return [currentList]

        include = helper(index + 1, currentList + text[index])
        
        exclude = helper(index + 1, currentList)
        
        return include + exclude
    
    return helper(0, '')

print(generateSubsequences("ab"))
# print(generateSubsequences("abc"))
```

# Problem 18 - Count Occurrences of a Character
Write a recursive function named countCharacter(text, target) that returns how many times target appears in text.

Examples
- text = "banana", target = "a" → 3
- text = "hello", target = "l" → 2
- text = "hello", target = "z" → 0

Constraints: Use recursion, without loops or built-in counting methods.

```python
def countCharacter(text, target):
    def helper(index):
        if index == len(text):
            return 0
        return 1 + helper(index + 1) if text[index] == target else helper(index + 1)
    return helper(0)

print(countCharacter('banana', 'a'))
print(countCharacter('hello', 'l'))
print(countCharacter('hello', 'z'))
```

# Problem 19 - Check Whether a Number Is a Power of Two
Write a recursive function named isPowerOfTwo(number) that returns True if the number is a power of two, otherwise False.

Examples
- number = 8 → True
- number = 16 → True
- number = 12 → False
- number = 1 → True

Constraint: Use recursion, without loops or bitwise operators.

```python
def isPowerOfTwo(number):
    if number == 1:
        return True
    if number % 2 != 0 or number <= 0:
        return False
    return isPowerOfTwo(number // 2)

print(isPowerOfTwo(8))
print(isPowerOfTwo(10))
```

# Problem 20 - Find the Number of Digits That Are Even
Write a recursive function named countEvenDigits(number) that returns the number of even digits in a non-negative integer. Count 0 as an even digit.

Examples
- number = 1234 → 2 (digits 2 and 4)
- number = 2468 → 4
- number = 1357 → 0
- number = 0 → 1

```python
def countEvenDigits(number):
    if number < 10:
        return 1 if number % 2 == 0 else 0
    return (1 if (number  % 10) % 2 == 0 else 0) + countEvenDigits(number // 10)

print(countEvenDigits(1234))
print(countEvenDigits(2468))
print(countEvenDigits(0))
```

# Problem 21 - Reverse a Linked List Recursively
Write a recursive Python function named reverseLinkedList(head) that reverses a singly linked list and returns the new head.
For this problem, each node has a .value and a .next attribute.

Constraints
- Use recursion, not loops.
- Reverse the links themselves; don't create a new linked list.
- Aim for O(n) time complexity.

```python
class LinkedList:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
def reverseLinkedList(head):
    if head is None or head.next is None:
        return head
    
    newHead = reverseLinkedList(head.next)
    
    head.next.next = head
    head.next = None
    return newHead
    
one = LinkedList(1)
two = LinkedList(2)
three = LinkedList(3)
four = LinkedList(4)

one.next = two
two.next = three
three.next = four

reverseLinkedList(one)
```

# Problem 22 - Find the Height of a Binary Tree
Write a recursive function named findTreeHeight(root) that returns the height of a binary tree, measured in the number of nodes on the longest path from the root to a leaf.

For an empty tree, return 0.

Examples

- Empty tree → 0
- A tree with only root 1 → 1
- Tree 1, with children 2 and 3, and node 2 having child 4 → 3

```python
class Tree:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
```

```python
def findTreeHeight(root):
    if root is None:
        return 0
    left = findTreeHeight(root.left)
    right = findTreeHeight(root.right)
    return 1 + max(left, right)
tree = Tree(1)
tree.left = Tree(2)
tree.right = Tree(3)
tree.left.left = Tree(4)
tree.left.right = Tree(5)

findTreeHeight(tree)
```
