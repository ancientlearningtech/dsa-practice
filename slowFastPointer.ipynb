```python
class Tree:
    def __init__(self, value):
        self.value = value
        self.next = None
    
firstNode = Tree(1)
secondNode = Tree(2)
thirdNode = Tree(3)
fourthNode = Tree(4)
fifthNode = Tree(5)

firstNode.next = secondNode
secondNode.next = thirdNode
thirdNode.next = fourthNode
fourthNode.next = fifthNode
fifthNode.next = None
```

# Problem 1 — Detect a Cycle in a Linked List

Given the head of a singly linked list, determine whether the linked list contains a cycle. # A cycle exists when following .next eventually causes us to revisit a node.

Example:

    1 → 2 → 3 → 4
            ↑   ↓
            ← ←

Output: True

```python
def hasCycle(head):
    if head is None:
        return False
    
    slow = head
    fast = head
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True

    return False

print(hasCycle(firstNode))
```

# Problem 2 — Find the Middle of a Linked List

Given the head of a singly linked list, return the middle node. If the list has an even number of nodes, return the second middle node.

Example 1

    1 → 2 → 3 → 4 → 5 → None

    Output: 3

```python
def findMiddleNode(head):
    if head is None:
        return None
    
    slow = head
    fast = head
    
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
    
    return slow
        
print(findMiddleNode(firstNode))
```

# Problem 3 — Happy Number

Given a positive integer n, repeatedly replace the number with the sum of the squares of its digits. If eventually the number becomes 1, it is a happy number. If the process enters a cycle that never reaches 1, it is not a happy number.

Example

    n = 19

Process:

    19
    ↓
    1² + 9² = 82
    ↓
    8² + 2² = 68
    ↓
    6² + 8² = 100
    ↓
    1² + 0² + 0² = 1

    Output: True

```python
# Helper function

def nextNumber(n):
    if n < 10:
        return n * n
    return nextNumber(n // 10) + pow(n % 10, 2)
```

```python
def isHappyNumber(n):
    slow = nextNumber(n)
    fast = nextNumber(nextNumber(n))
    
    while True:
        if slow == 1:
            return True
        elif slow == fast:
            return False
        slow = nextNumber(slow)
        fast = nextNumber(nextNumber(fast))

print(isHappyNumber(19))
```

# Problem 4 — Find the Start of a Cycle

Example:

    1 → 2 → 3 → 4
        ↑       ↓
        └───────┘

The cycle starts at node 2.

    Output: Node(2)

```python
def findCycleStart(head):
    slow = head
    fast = head
    
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            break
    else:
        return None
    
    fast = head
    while slow != fast:
        slow = slow.next
        fast = fast.next
    
    return slow

print(findCycleStart(firstNode))
```

# Problem 5 - Remove the Nth Node From the End of a Linked List

Given the head of a singly linked list, remove the Nth node from the end of the list and return the new head.

Example

    Input:  1 → 2 → 3 → 4 → 5,  n = 2

    Output: 1 → 2 → 3 → 5

```python
def removeNthNode(head, n):
    dummy = Tree(0)
    dummy.next = head
    slow = dummy
    fast = head

    for _ in range(n):
        fast = fast.next
        
    while fast is not None:
        slow = slow.next
        fast = fast.next
        
    slow.next = slow.next.next

    return dummy.next
    
firstNode = Tree(1)
secondNode = Tree(2)
thirdNode = Tree(3)
fourthNode = Tree(4)
fifthNode = Tree(5)

firstNode.next = secondNode
secondNode.next = thirdNode
thirdNode.next = fourthNode
fourthNode.next = fifthNode
fifthNode.next = None

print(removeNthNode(firstNode, 2))
```

# Problem 6 - Palindrome Linked List

Given the head of a singly linked list, determine whether the linked list is a palindrome.

Example

    1 → 2 → 2 → 1

    Output: True

```python
def isPalindrome(head):
    if head is None or head.next is None:
        return True
    
    slow = head
    fast = head
    
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
    
    previous = None
    current = slow
    
    while current is not None:
        nextNode = current.next
        current.next = previous
        previous = current
        current = nextNode
      
    secondHalf = previous
    firstHalf = head
      
    while secondHalf is not None:
        if firstHalf.value != secondHalf.value:
            return False

        firstHalf = firstHalf.next
        secondHalf = secondHalf.next
        
    return True

firstNode = Tree(1)
secondNode = Tree(2)
thirdNode = Tree(3)
fourthNode = Tree(4)
fifthNode = Tree(5)

firstNode.next = secondNode
secondNode.next = thirdNode
thirdNode.next = fourthNode
fourthNode.next = fifthNode
fifthNode.next = None

print(isPalindrome(firstNode))
```

# Problem 7 - Find the Length of the Cycle

You are given the head of a singly linked list. The list may or may not contain a cycle. Return the number of nodes in the cycle. If there is no cycle, return 0.

Example:

    1 → 2 → 3 → 4 → 5
            ↑       ↓
            ← ← ← ←

    The cycle is: 3 → 4 → 5 → 3

    Output: 3

```python
def findLengthOfCycle(head):
    if head is None:
        return 0
    
    slow = head
    fast = head
    
    # Detect the meeting node of the cycle.
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            break
    else:
        return 0
    
    # Count the cycle
    count = 1
    current = slow.next
    
    while current != slow:
        current = current.next
        count += 1
    
    return count

firstNode = Tree(1)

secondNode = Tree(2)
thirdNode = Tree(3)
fourthNode = Tree(4)
fifthNode = Tree(5)

firstNode.next = secondNode
secondNode.next = thirdNode
thirdNode.next = fourthNode
fourthNode.next = fifthNode
fifthNode.next = thirdNode

print(findLengthOfCycle(firstNode))
```
