# Problem 1 — Detect a Cycle in a Linked List

Given the head of a singly linked list, determine whether the linked list contains a cycle. # A cycle exists when following .next eventually causes us to revisit a node.

Example:

    1 → 2 → 3 → 4
            ↑   ↓
            ← ←

Output: True

```python
def hasCycle(head):
    if head is None:
        return False
    
    slow = head
    fast = head
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True

    return False

print(hasCycle(firstNode))
```

# Problem 2 — Find the Middle of a Linked List

Given the head of a singly linked list, return the middle node. If the list has an even number of nodes, return the second middle node.

Example 1

    1 → 2 → 3 → 4 → 5 → None

    Output: 3

```python
def findMiddleNode(head):
    if head is None:
        return None
    
    slow = head
    fast = head
    
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
    
    return slow
        
print(findMiddleNode(firstNode))
```
