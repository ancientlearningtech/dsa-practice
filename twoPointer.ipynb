# Problem 1: Check for a Pair With a Target Sum

You are given a sorted array of integers and a target number. Determine whether there are two different elements in the array whose sum equals the target. Return True if such a pair exists; otherwise, return False.

Example:

numbers = [1, 2, 3, 4, 6]

target = 6

Output: True

```python
def hasPairWithTargetSum(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left < right:
        if numbers[left] + numbers[right] > target:
            right -= 1
        elif numbers[left] + numbers[right] < target:
            left += 1
        else:
            return True
    
    return False

print(hasPairWithTargetSum([1,2,3,4,6], 6))
print(hasPairWithTargetSum([1, 2, 4, 4, 6], 8))
```

# Problem 2: Reverse an Array In-Place

Given an array of integers, reverse the array in-place using two pointers. Do not create a new array. Modify the original array directly.

Example:

    numbers = [1, 2, 3, 4, 5]

    Output:[5, 4, 3, 2, 1]

```python
def reverseArray(numbers):
    left = 0
    right = len(numbers) - 1
    
    while left < right:
        numbers[left], numbers[right] = numbers[right], numbers[left]
        left += 1
        right -= 1
    
    return numbers

print(reverseArray([1,2,3,4,5]))
print(reverseArray([1,2,3,4,5,6]))
```

# Problem 3: Valid Palindrome

Given a string, determine whether it is a palindrome.

text = "A man, a plan, a canal: Panama"

Output: True

```python
def isPalindrome(text):
    left = 0
    right = len(text) - 1
    
    while left < right:
        while left < right and not text[left].isalnum():
            left += 1
        while left < right and not text[right].isalnum():
            right -= 1
        if text[left].lower() != text[right].lower():
            return False
        left += 1
        right -= 1
        
    return True

print(isPalindrome('Racecar'))
print(isPalindrome('racecar'))
print(isPalindrome("A man, a plan, a canal: Panama"))
```

# Problem 4: Remove Duplicates From a Sorted Array

Given a sorted array, remove duplicates in-place so that each value appears only once. Return the number of unique elements.

Example:

    numbers = [1, 1, 2, 2, 3]

After modifying the array:

    [1, 2, 3, 2, 3]

Only the first 3 positions matter.

Return: 3

```python
def removeDuplicates(numbers):
    slow = 0
    fast = 1
    
    while fast < len(numbers):
        if numbers[slow] != numbers[fast]:
            slow += 1
            numbers[slow] = numbers[fast]

        fast += 1
    return numbers[:slow + 1]
        
print(removeDuplicates([1,1,2,2,3]))
print(removeDuplicates([1, 1, 1, 2, 2, 3, 3, 4]))
```

# Problem 5: Move Zeroes

Given an array of integers, move all 0s to the end of the array while maintaining the relative order of the non-zero elements. Do this in-place.

Examples

    numbers = [0, 1, 0, 3, 12]

    Output: [1, 3, 12, 0, 0]

```python
def moveZeroes(numbers):
    slow = 0
    fast = 0
    
    while fast < len(numbers):
        if numbers[fast] != 0:
            numbers[slow], numbers[fast] = numbers[fast], numbers[slow]
            slow += 1
        fast += 1
    return numbers
print(moveZeroes([0,1,0,3,12]))
print(moveZeroes([1,0,1,3,12]))
```

 # Problem 6 — Two Sum II

 Given a sorted array, return the indices of two numbers whose sum equals the target.

Example:

    numbers = [2, 7, 11, 15]
    target = 9

Expected: [1, 2]

```python
def findTwoSumIndices(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left < right:
        if numbers[left] + numbers[right] > target:
            right -= 1
        elif numbers[left] + numbers[right] < target:
            left += 1
        else:
            return [left + 1, right + 1]
    
    return []

print(findTwoSumIndices([2,7,11,15], 9))
print(findTwoSumIndices([1, 2, 3, 4, 6], 7))
```

# Problem 7: Container With Most Water

You are given an array of non-negative integers representing vertical lines. Find two lines that, together with the x-axis, form a container that holds the maximum amount of water. Return the maximum area.

Example

    heights = [1, 8, 6, 2, 5, 4, 8, 3, 7]

    Output: 49

```python
def findMaxWaterContainer(heights):
    left = 0
    right = len(heights) - 1
    maximumWater = 0
    
    while left < right:
        currentVolume = min(heights[left], heights[right]) * (right - left)
        maximumWater = max(maximumWater, currentVolume)
        if heights[left] >= heights[right]:
            right -= 1
        else:
            left += 1
    return maximumWater

print(findMaxWaterContainer([1, 8, 6, 2, 5, 4, 8, 3, 7]))
print(findMaxWaterContainer([2, 3, 4, 5, 1]))
```

# Problem 8: 3Sum

Given an integer array, find all unique triplets [a, b, c] such that: a + b + c = 0

Example:

    numbers = [-1, 0, 1, 2, -1, -4]

    Output: [[-1, -1, 2], [-1, 0, 1]]

```python
def findThreeSum(numbers):
    numbers.sort()
    result = []
    
    for i in range(len(numbers) - 2):
        if i > 0 and numbers[i] == numbers[i - 1]:
            continue
        left = i + 1
        right = len(numbers) - 1
        
        while left < right:
            currentSum = numbers[i] + numbers[left] + numbers[right]

            if currentSum == 0:
                result.append([numbers[i], numbers[left], numbers[right]])
                left += 1
                right -= 1
                
                while left < right and numbers[left] == numbers[left - 1]:
                    left += 1

                while left < right and numbers[right] == numbers[right + 1]:
                    right -= 1

            elif currentSum > 0:
                right -= 1

            else:
                left += 1
        
    return result

print(findThreeSum([-1,0,1,2,-1,-4]))
```
