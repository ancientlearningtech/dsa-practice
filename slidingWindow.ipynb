# Problem 1: Maximum Sum of a Subarray of Size K

Given an array of integers and an integer k, find the maximum sum of any contiguous subarray of exactly k elements.

Example 1:

numbers = [2, 1, 5, 1, 3, 2]

k = 3

The possible windows are:

    [2, 1, 5] → 8
    [1, 5, 1] → 7
    [5, 1, 3] → 9
    [1, 3, 2] → 6

Output = 9

```python
def maximumSubarraySum(numbers, k):
    left = 0
    right = k
    currentSum = sum(numbers[:k])
    maximumSum = currentSum
    
    while left < len(numbers) - k:
        currentSum += numbers[right] - numbers[left]
        maximumSum = max(maximumSum, currentSum)
        left += 1
        right += 1
    
    return maximumSum
    
print(maximumSubarraySum([2,1,5,1,3,2], 3))
print(maximumSubarraySum([2, 3, 4, 1, 5], 2))
print(maximumSubarraySum([5, -2, 3, 1], 2))
```

# Problem 2: Average of Every Subarray of Size K

Given an array of integers and an integer k, return the average of every contiguous subarray of exactly k elements.

Example 1

numbers = [1, 3, 2, 6, -1, 4, 1, 8, 2]

k = 5

The windows are:

    [1, 3, 2, 6, -1] → 11 / 5 = 2.2
    [3, 2, 6, -1, 4] → 14 / 5 = 2.8
    [2, 6, -1, 4, 1] → 12 / 5 = 2.4
    [6, -1, 4, 1, 8] → 18 / 5 = 3.6
    [-1, 4, 1, 8, 2] → 14 / 5 = 2.8

Result: [2.2, 2.8, 2.4, 3.6, 2.8]

```python
def findAverages(numbers, k):
    left = 0
    right = k
    currentSum = sum(numbers[:k])
    subarrayAverage = [currentSum / k]
    
    while left < len(numbers) - k:
        currentSum += numbers[right] - numbers[left]
        subarrayAverage.append(currentSum / k)
        left += 1
        right += 1
    
    return subarrayAverage

print(findAverages([1, 3, 2, 6, -1, 4, 1, 8, 2], 5))
print(findAverages([2, 4, 6, 8], 2))
```

# Problem 3: Maximum Sum of Subarray of Size K

Let's increase the difficulty slightly.

Given an array of positive integers and an integer k, find the smallest contiguous subarray whose sum is at least k.

numbers = [2, 1, 5, 2, 3, 2], k = 7

Smallest subarray with sum >= 7:

[5, 2] → sum = 7

Answer = 2

```python
def findSmallestSubarray(numbers, k):
    left = 0
    currentSum = 0
    minimumLength = float('inf')
    
    for right in range(len(numbers)):
        currentSum += numbers[right]
        
        while currentSum >= k:
            currentLength = right - left + 1
            minimumLength = min(minimumLength, currentLength)
            
            currentSum -= numbers[left]
            left += 1
            
    if minimumLength == float('inf'):
        return 0

    return minimumLength

print(findSmallestSubarray([2, 1, 5, 2, 3, 2], 7))
print(findSmallestSubarray([2, 3, 1, 2, 4, 3], 7))
```

# Problem 4 — Longest Subarray with Sum ≤ K

Given an array of positive integers and an integer k, find the length of the longest contiguous subarray whose sum is less than or equal to k.

Example 1

numbers = [2, 1, 5, 2, 3, 2], k = 7

Some valid windows:

    [2, 1, 5]       → sum = 8 ❌
    [1, 5]          → sum = 6  ✓
    [5, 2]          → sum = 7  ✓
    [2, 3, 2]       → sum = 7  ✓
    [1, 5, 2]       → sum = 8 ❌

The longest valid window has length: 3

```python
def findLongestSubarray(numbers, k):
    left = 0
    maximumLength = 0
    currentSum = 0
    
    for right in range(len(numbers)):
        currentSum += numbers[right]
        while currentSum > k:
            currentSum -= numbers[left]
            left += 1

        currentLength = right - left + 1
        maximumLength = max(maximumLength, currentLength)
            
    return maximumLength

print(findLongestSubarray([2, 1, 5, 2, 3, 2], 7))
print(findLongestSubarray([2, 3, 1, 2, 4, 3], 7))
```

# Problem 5 — Longest Substring with At Most K Distinct Characters

Given a string text and an integer k, find the length of the longest substring containing at most k distinct characters.

A substring must be contiguous.

Example 1

text = "araaci"

k = 2

The longest valid substring is: "araa"

It contains: a, r

which is 2 distinct characters.

Answer: 4

```python
def findLongestSubstring(text, k):
    left = 0
    maximumLength = 0
    kStringCount = dict()
    
    for right in range(len(text)):
        kStringCount[text[right]] = kStringCount.get(text[right], 0) + 1
        while len(kStringCount) > k:
            kStringCount[text[left]] -= 1
            if kStringCount[text[left]] == 0:
                del kStringCount[text[left]]
            left += 1
            
        maximumLength = max(maximumLength, right - left + 1)
    
    return maximumLength
        
print(findLongestSubstring('araaci', 2))
print(findLongestSubstring('cbbebi', 3))
```

# Problem 6 — Longest Substring with No Repeating Characters

Given a string text, find the length of the longest substring that contains no repeated characters.

A substring must be contiguous.

Example:

text = "abcabcbb"

Longest substring without repeating characters: "abc"

Answer: 3

```python
def findLongestUniqueSubstring(text):
    left = 0
    maximumLength = 0
    sourceDict = dict()
    
    for right in range(len(text)):
        sourceDict[text[right]] = sourceDict.get(text[right], 0) + 1

        while sourceDict[text[right]] > 1: # can also use `while len(sourceDict) < right - left + 1:`
            sourceDict[text[left]] -= 1
            if sourceDict[text[left]] == 0:
                del sourceDict[text[left]]
            left += 1
                        
        maximumLength = max(maximumLength, right - left + 1)

    return maximumLength

print(findLongestUniqueSubstring("abcabcbb"))
```

# Problem 7 — Maximum Number of Vowels in a Substring of Size K

Given a string text and an integer k, find the maximum number of vowels (a, e, i, o, u) in any contiguous substring of exactly k characters.

Example 1:

text = "abciiidef"

k = 3

Windows include:

    "abc" → 1 vowel
    "bci" → 1 vowel
    "cii" → 2 vowels
    "iii" → 3 vowels  ← maximum

Answer: 3

```python
def findMaximumVowels(text, k):
    vowels = {'a', 'e', 'i', 'o', 'u'}

    currentVowelsCount = 0

    for character in text[:k]:
        if character in vowels:
            currentVowelsCount += 1

    maximumVowels = currentVowelsCount

    left = 0

    for right in range(k, len(text)):
        if text[left] in vowels:
            currentVowelsCount -= 1

        if text[right] in vowels:
            currentVowelsCount += 1

        maximumVowels = max(maximumVowels, currentVowelsCount)

        left += 1

    return maximumVowels


print(findMaximumVowels("abciiidef", 3))  # 3
print(findMaximumVowels("aeiou", 2))      # 2
print(findMaximumVowels("leetcode", 3))   # 2
```

# Problem 8 — Permutation in String

Given two strings text and pattern, return True if text contains a substring that is a permutation (rearrangement) of pattern. Otherwise, return False.

Examples

text = "oidbcaf"

pattern = "abc"

Output: True

```python
def checkPermutation(text, pattern):
    targetDict = dict()
    for character in pattern:
        targetDict[character] = targetDict.get(character, 0) + 1
    sourceDict = dict()
    for character in text[:len(pattern)]:
        sourceDict[character] = sourceDict.get(character, 0) + 1
        
    if sourceDict == targetDict:
        return True
    left = 0
    for right in range(len(pattern), len(text)):
        if sourceDict == targetDict:
            return True
        sourceDict[text[right]] = sourceDict.get(text[right], 0) + 1
        sourceDict[text[left]] -= 1
        if sourceDict[text[left]] == 0:
            del sourceDict[text[left]]
        
        left += 1
        
        if sourceDict == targetDict:
            return True
    return False
    
print(checkPermutation('oidbcaf', 'abc')) # True
print(checkPermutation('bcdxabcdy', 'bcdyabcd'))  # False
print(checkPermutation('odicf', 'dc')) # False
```

Problem 9 — Longest Repeating Character Replacement

Given a string containing uppercase English letters and an integer k, you may replace at most k characters.

Return the length of the longest substring that can be made to contain only one repeated character after at most k replacements.

Example:
text = "ABAB"
k = 2

Output: 4

We can change:

    A B A B
    ↓   ↓
    A A A A

Answer: 4.

```python
def findLongestRepeatingSubstring(text, k):
    left = 0
    mostFrequentCount = 0
    maximumLength = 0
    
    sourceDict = dict()
    
    for right in range(len(text)):
        sourceDict[text[right]] = sourceDict.get(text[right], 0) + 1
        mostFrequentCount = max(mostFrequentCount, sourceDict[text[right]])
        
        while (right - left + 1) - mostFrequentCount > k:
            sourceDict[text[left]] -= 1
            if sourceDict[text[left]] == 0:
                del sourceDict[text[left]]
            left += 1
        maximumLength = max(maximumLength, right - left + 1)
    return maximumLength
        
print(findLongestRepeatingSubstring("ABAB", 2)) # 4
print(findLongestRepeatingSubstring("AABABBA", 1))    # 4
```
