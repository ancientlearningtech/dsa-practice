# Problem 1 — Meeting Rooms

You are given a list of meeting intervals:

    meetings = [[0,30], [5,10], [15,20]]

    Each interval represents: [startTime, endTime]

Determine whether one person can attend all meetings without any two meetings overlapping.

Return:

    True if all meetings can be attended

    False if at least two meetings overlap

Example:

    Input: [[0,30], [5,10], [15,20]]

    Output False

```python
def canAttendAllMeetings(meetings):
    meetings.sort(key=lambda meeting: meeting[0])
    
    for i in range(1, len(meetings)):
        if meetings[i][0] < meetings[i - 1][1]:
            return False
    return True
    
print(canAttendAllMeetings([[0,30], [5,10], [15,20]]))
print(canAttendAllMeetings([[7,10], [2,4]]))
print(canAttendAllMeetings([[1,5], [5,8]]))
```

# Problem 2 — Non-Overlapping Intervals

You are given a list of intervals:  Return the minimum number of intervals you need to remove so that the remaining intervals do not overlap.

Example:

    Input: [[1,2], [2,3], [3,4], [1,3]]

    Output: 1

```python
def eraseOverlapIntervals(intervals):
    intervals.sort(key=lambda interval:interval[1])
    removedCount = 0
    previousEnd = intervals[0][1]
    
    i = 1
    while i < len(intervals):
        currentStart = intervals[i][0]
        if previousEnd > currentStart:
            removedCount += 1
        else:
            previousEnd = intervals[i][1]
        i += 1

    return removedCount

print(eraseOverlapIntervals([[1,2], [2,3], [3,4], [1,3]]))
print(eraseOverlapIntervals([[1,4], [2,3], [3,5]]))
```

# Problem 3 - Minimum Number of Arrows to Burst Balloons

You are given a list of balloons represented by their horizontal diameter: [startX, endX]

A balloon occupies every point between startX and endX. You can shoot an arrow vertically from any x coordinate. An arrow bursts every balloon whose interval contains that x coordinate. Return the minimum number of arrows needed to burst all balloons.

Example:

    Input: [[10,16], [2,8], [1,6], [7,12]]

    Output: 2

```python
def findMinArrowShots(points):
    points.sort(key=lambda point:point[1])
    print(points)
    arrowPosition = points[0][1]
    requiredArrows = 1
    
    for i in range(1, len(points)):
        currentStart = points[i][0]
        currentEnd = points[i][1]
        if currentStart > arrowPosition:
            requiredArrows += 1
            arrowPosition = currentEnd
    
    return requiredArrows
    
print(findMinArrowShots([[10,16], [2,8], [1,6], [7,12]])) # 2
print(findMinArrowShots([[1,2], [3,4], [5,6], [7,8]])) # 4
print(findMinArrowShots([[1,10], [2,5], [3,7], [6,9]])) # 1
```

# Problem 4 - Interval List Intersections

You are given two lists of closed intervals. Each list is sorted by starting time non-overlapping within itself. Return all intervals where the two lists overlap.

Example

    firstList  = [[0,2], [5,10], [13,23], [24,25]]

    secondList = [[1,5], [8,12], [15,24], [25,26]]

    Expected: [[1,2], [5,5], [8,10], [15,23], [24,24], [25,25]]

```python
def findIntervalIntersections(firstList, secondList):
    i = 0
    j = 0
    result = []
    
    while i < len(firstList) and j < len(secondList):
        firstStart = firstList[i][0]
        firstEnd = firstList[i][1]
        
        secondStart = secondList[j][0]
        secondEnd = secondList[j][1]
        
        overlapStart  = max(firstStart, secondStart)
        overlapEnd = min(firstEnd, secondEnd)
        
        if overlapStart <= overlapEnd:
            result.append([overlapStart, overlapEnd])
            
        if firstEnd < secondEnd:
            i += 1
        else:
            j += 1
    
    return result
        
print(findIntervalIntersections([[0,2], [5,10], [13,23], [24,25]], 
                                [[1,5], [8,12]])) # [[1,2], [5,5], [8,10]]

print(findIntervalIntersections([[0,2], [5,10], [13,23], [24,25]],
                                [[1,5], [8,12], [15,24], [25,26]]))
```

# Problem 5 — Employee Free Time

You are given the working schedules of multiple employees. Each employee's schedule contains non-overlapping intervals, and each employee's intervals are sorted by start time. Find the intervals when all employees are free.

Example:

    schedule = [
        [[1, 3], [6, 7]], 
        [[2, 4]], 
        [[2, 5], [9, 12]]
    ]

    output: [[5, 6], [7, 9]]

```python
def flattenList(myList, result):
    if not isinstance(myList[0], list):
        result.append(myList)
        return
    
    for item in myList:
        flattenList(item, result)
        
    return result
```

```python
def findEmployeeFreeTime(schedule):
    # flatten the input list
    flattenSchedule = flattenList(schedule, [])
    
    # sort the flattened input by start time
    flattenSchedule.sort(key=lambda flat: flat[0])
    
    # Merge overlapping intervals
    merged = [flattenSchedule[0].copy()]
        
    for i in range(1, len(flattenSchedule)):
        currentStart = flattenSchedule[i][0]
        currentEnd = flattenSchedule[i][1]
        
        previousEnd = merged[-1][1]
        
        if currentStart <= previousEnd:
            merged[-1][1] = max(previousEnd, currentEnd)
        else:
            merged.append(flattenSchedule[i].copy())
    
    # generate free interval
    freeInterval = []
    for i in range(1, len(merged)):
        freeInterval.append([merged[i - 1][1], merged[i][0]])
    
    return freeInterval

print(findEmployeeFreeTime([[[1, 3], [6, 7]], [[2, 4]], [[2, 5], [9, 12]]]))
# print(findEmployeeFreeTime())
```

# Problem 6 — Remove Covered Intervals

Given a list of intervals, remove every interval that is completely covered by another interval. Return the number of intervals that remain.

Example:

    intervals = [[1,4], [3,6], [2,8]]

    Output: 2

```python
def removeCoveredIntervals(intervals):
    intervals.sort(key=lambda interval: (interval[0], -interval[1]))
    count = 0
    farthestEnd = 0
    for start, end in intervals:
        if end > farthestEnd:
            count += 1
            farthestEnd = end
            
    return count        
    
print(removeCoveredIntervals([[1,4],[3,6],[2,8]])) # 2
print(removeCoveredIntervals([[1,2], [2,3], [3,4]])) # 3
print(removeCoveredIntervals([[1,3], [1,5], [1,8]])) # 1
```
