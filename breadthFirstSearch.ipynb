```python
class Tree:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
```

# Problem 1 — Binary Tree Level Order Traversal

Given the root of a binary tree, return the values of its nodes level by level.

Example

              1
            /   \
           2     3
          / \     \
         4   5     6

Expected output:

    [
        [1],
        [2, 3],
        [4, 5, 6]
    ]

```python
from collections import deque
def levelOrderTraversal(root):
    if root is None:
        return []
    queue = deque([root])
    result = []
    
    while queue:
        levelSize = len(queue)
        currentLevel = []
        for _ in range(levelSize):
            node = queue.popleft()
            currentLevel.append(node.value)
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
        result.append(currentLevel)
    return result

new = Tree(1)
new.left = Tree(2)
new.right = Tree(3)
new.left.left = Tree(4)
new.left.right = Tree(5)
new.right.right = Tree(6)
print(levelOrderTraversal(new))
```

# Problem 2 — Binary Tree Rightmost Node at Each Level

Example

              1
            /   \
           2     3
          / \     \
         4   5     6
              \
               7

    Expected output: [1, 3, 6, 7]

```python
def findRightmostNodes(root):
    if root is None:
        return []
    queue = deque([root])
    result = []
    
    while queue:
        levelSize = len(queue)
        for _ in range(levelSize):
            node = queue.popleft()
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
        
        result.append(node.value)
        
    return result
   
new = Tree(1)
new.left = Tree(2)
new.right = Tree(3)
new.left.left = Tree(4)
new.left.right = Tree(5)
new.right.right = Tree(6)
new.left.right.right = Tree(7)
print(findRightmostNodes(new))
```

# Problem 3 — Find the Average of Each Level

Given the root of a binary tree, return the average value of the nodes at each level, from top to bottom.

Example:

              1
            /   \
           2     3
          / \     \
         4   5     6
    
    Output: [1.0, 2.5, 5.0]

```python
from collections import deque
def findLevelAverages(root):
    if root is None:
        return []
    queue = deque([root])
    result = []
    
    while queue:
        levelSize = len(queue)
        currentSum = 0
        for _ in range(levelSize):
            node = queue.popleft()
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
            currentSum += node.value
        result.append(currentSum / levelSize)

    return result
new = Tree(1)
new.left = Tree(2)
new.right = Tree(3)
new.left.left = Tree(4)
new.left.right = Tree(5)
new.right.right = Tree(6)
print(findLevelAverages(new))
```

# Problem 4 — Minimum Depth of a Binary Tree

Given the root of a binary tree, find its minimum depth. The minimum depth is the number of nodes along the shortest path from the root to a leaf node. A leaf node is a node that has no left or right child.

Example

Binary tree

                1
              /   \
             2     3
            /
           4
  
The paths to leaf nodes are:

    1 → 3 — depth 2

    1 → 2 → 4 — depth 3

    Expected output: 2

```python
def findMinimumDepth(root):
    if root is None:
        return 0
    queue = deque([root])
    currentDepth = 0
    
    while queue:
        levelSize = len(queue)
        currentDepth += 1
        for _ in range(levelSize):
            node = queue.popleft()
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
            if node.left is None and node.right is None:
                return currentDepth
        
    return currentDepth
    
new = Tree(1)
new.left = Tree(2)
new.right = Tree(3)
# new.left.left = Tree(4)
new.right.right = Tree(5)
# new.left.right = Tree(5)
# new.right.right = Tree(6)
print(findMinimumDepth(new))
```

# Problem 5 - Binary Tree Zigzag Level Order Traversal

Given the root of a binary tree, return its node values level by level, but alternate the direction at each level:

Example:

              1
            /   \
           2     3
          / \   / \
         4   5 6   7

Expected output:

        [
            [1],
            [3, 2],
            [4, 5, 6, 7]
        ]

```python
from collections import deque
def zigzagLevelOrder(root):
    if root is None:
        return []
    queue = deque([root])
    result = []
    reverseFlag = False
    
    while queue:
        levelSize = len(queue)
        currentLevel = []
        
        for _ in range(levelSize):
            node = queue.popleft()
            currentLevel.append(node.value)
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
        if reverseFlag:
            currentLevel.reverse()
        result.append(currentLevel)
        reverseFlag = not reverseFlag
    
    return result

new = Tree(1)
new.left = Tree(2)
new.right = Tree(3)
new.left.left = Tree(4)
new.left.right = Tree(5)
new.right.left = Tree(6)
new.right.right = Tree(7)
print(zigzagLevelOrder(new))
```

# Problem 6: Binary Tree Level Order Successor

Given the root of a binary tree and a target value, return the next node that appears after the target in level-order traversal. If the target is the last node in the traversal, return None.

Example:

          1
        /   \
       2     3
      / \     \
     4   5     6

Level-order traversal is: [1, 2, 3, 4, 5, 6]

```python
from collections import deque
def findLevelOrderSuccessor(root, target):
    if root is None:
        return None
    queue = deque([root])
    flag = False
    
    while queue:
        levelSize = len(queue)
        for _ in range(levelSize):
            node = queue.popleft()
            if flag:
                return node.value
            if target == node.value:
                flag = not flag
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
            
    return None

new = Tree(1)
new.left = Tree(2)
new.right = Tree(3)
new.left.left = Tree(4)
new.left.right = Tree(5)
new.right.left = Tree(6)
new.right.right = Tree(7)
print(findLevelOrderSuccessor(new, 1))
```

# Problem 7 - Connect Level-Order Siblings

Given the root of a binary tree, connect each node to the next node on the same level using a .next pointer.

Example

          1
        /   \
       2     3
      / \     \
     4   5     6

    Level 1: 1 → None

    Level 2: 2 → 3 → None

    Level 3: 4 → 5 → 6 → None

```python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        self.next = None
```

```python
from collections import deque
def connectLevelOrderSiblings(root):
    if root is None:
        return None
    queue = deque([root])

    while queue:
        levelSize = len(queue)
        previousNode = None
        for _ in range(levelSize):
            node = queue.popleft()
            if previousNode is not None:
                previousNode.next = node
            previousNode = node
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
        node.next = None

    return root
```

# Problem 8 - Find the Level with the Maximum Sum.

Given the root of a binary tree, return the level number that has the largest sum of node values. The root is at level 1. If multiple levels have the same sum, return the smallest level number.

Example

          1          ← Level 1: sum = 1
        /   \
       2     3       ← Level 2: sum = 5
      / \     \
     4   5     6     ← Level 3: sum = 15

    Expected output: 3

```python
from collections import deque
def findLevelWithMaximumSum(root):
    if root is None:
        return 0
    queue = deque([root])
    maximumSum = float('-inf')
    result = 0
    level = 1
    
    while queue:
        levelSize = len(queue)
        currentSum = 0

        for _ in range(levelSize):
            node = queue.popleft()
            currentSum += node.value
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
        if currentSum > maximumSum:
            maximumSum = currentSum
            result = level
        level += 1
    
    return result

new = Tree(1)
new.left = Tree(2)
new.right = Tree(3)
new.left.left = Tree(4)
new.left.right = Tree(5)
new.right.right = Tree(6)
# new.right.right = Tree(7)
print(findLevelWithMaximumSum(new))
```

# Problem 9 - Find the Maximum Value at Each Level

Given the root of a binary tree, return a list containing the maximum node value at each level, from top to bottom.

Example

          1
        /   \
       2     3
      / \   / \
     4   5 6   7

    Output: [1, 3, 7]

```python
from collections import deque
def findMaximumValueAtEachLevel(root):
    if root is None:
        return None
    queue = deque([root])
    result = []
    
    while queue:
        levelSize = len(queue)
        maximumValue = float('-inf')
        
        for _ in range(levelSize):
            node = queue.popleft()
            maximumValue = max(node.value, maximumValue)
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
        
        result.append(maximumValue)
    
    return result

new = Tree(1)
new.left = Tree(2)
new.right = Tree(3)
new.left.left = Tree(4)
new.left.right = Tree(5)
new.right.right = Tree(6)
new.right.right = Tree(7)
print(findMaximumValueAtEachLevel(new))
```

# Problem 10 - Binary Tree Level Order Traversal in Reverse

Given the root of a binary tree, return its level-order traversal from the bottom level to the top level. Within each level, keep the nodes in their normal left-to-right order.

Example:

          1
        /   \
       2     3
      / \   / \
     4   5 6   7

    Output: [[4, 5, 6, 7], [2, 3], [1]]

```python
from collections import deque
def reverseLevelOrderTraversal(root):
    if root is None:
        return []
    queue = deque([root])
    result = []
    
    while queue:
        levelSize = len(queue)
        currentLevel = []
        
        for _ in range(levelSize):
            node = queue.popleft()
            currentLevel.append(node.value)
            
            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)
        
        result.append(currentLevel)
        
    result.reverse()
    return result

new = Tree(1)
new.left = Tree(2)
new.right = Tree(3)
new.left.left = Tree(4)
new.left.right = Tree(5)
new.right.left = Tree(6)
new.right.right = Tree(7)
print(reverseLevelOrderTraversal(new))
```

# Problem 11 - Symmetric Tree

Given the root of a binary tree, determine whether the tree is symmetric around its center — meaning its left and right subtrees are mirror images of each other.
Two subtrees are mirror images when:
- Their root values are equal.
- The left child of one node matches the right child of the other.
- The right child of one node matches the left child of the other.
- This mirrored relationship holds for all corresponding nodes, including missing children.

```python
from collections import deque
def isSymmetric(root):
    if root is None:
        return True
    queue = deque([(root.left, root.right)])
    
    while queue:
        leftNode, rightNode = queue.popleft()
        if leftNode is None and rightNode is None:
            continue
        if leftNode is None or rightNode is None:
            return False
        if leftNode.value != rightNode.value:
            return False
        
        queue.append((leftNode.left, rightNode.right))
        queue.append((leftNode.right, rightNode.left))
    
    return True

root = Tree(1)
root.left = Tree(2)
root.right = Tree(2)
root.left.left = Tree(3)
root.left.right = Tree(4)
root.right.left = Tree(4)
root.right.right = Tree(3)
print(isSymmetric(root))
```
