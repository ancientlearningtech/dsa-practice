# Problem 1 - Generate All Subsets
Given a list of distinct integers, return all possible subsets, including the empty subset and the original list.

    For example: numbers = [1, 2, 3]

    Output: [[],[1],[2],[3],[1, 2],[1, 3],[2, 3],[1, 2, 3]]

```python
def generateSubsets(numbers):
    def helper(index, currentList):
        if index == len(numbers):
            return [currentList]
        include = helper(index + 1, currentList + [numbers[index]])
        exclude = helper(index + 1, currentList)
        return include + exclude
    
    return helper(0, [])

print(generateSubsets([1,2,3]))
```
